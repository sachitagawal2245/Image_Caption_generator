```python
import os

images = os.listdir("images")
print(len(images))  
```

```python
import random

random.shuffle(images)
train_size = int(0.7 * len(images))
val_size   = int(0.15 * len(images))
train_imgs = images[:train_size]
val_imgs   = images[train_size:train_size+val_size]
test_imgs  = images[train_size+val_size:]
```

```python
def save_split(file_path, data):
    with open(file_path, "w") as f:
        for item in data:
            f.write(item + "\n")

save_split("train.txt", train_imgs)
save_split("val.txt", val_imgs)
save_split("test.txt", test_imgs)
```
